```python
def rebin(arr, new_bins):
    """
    Rebin each L1 vector in arr from old_bins → new_bins by averaging the bins.
    arr: shape (n_real, old_bins)
    """
    arr = np.asarray(arr, dtype=float)
    old_bins = arr.shape[1]
    factor = old_bins // new_bins

    # reshape and average blocks
    return arr[:, :factor*new_bins].reshape(arr.shape[0], new_bins, factor).mean(axis=2)

new_bins = 100
fid_arr_rebinned = rebin(fid_arr, new_bins)

Cov_fid = np.cov(fid_arr_rebinned, rowvar=False)
eps = 1e-6 * np.mean(np.diag(Cov_fid))
Cov_fid_reg = Cov_fid #+ eps * np.eye(Cov_fid.shape[0])

fid_trace = np.trace(Cov_fid_reg)
fid_frob  = np.linalg.norm(Cov_fid_reg, "fro")

trace_ratios = []
frob_ratios = []
nreal_list = []

sim_reb = []
for i in range(len(cg_l1_kappa_realizations)):

    arr = np.asarray(cg_l1_kappa_realizations[i], dtype=float)

    # # clean up dtype object inconsistencies
    # if arr.ndim != 2:
    #     arr = np.vstack(arr)

    nreal = arr.shape[0]
    nreal_list.append(nreal)

    arr_rebinned = rebin(arr, new_bins)
    sim_reb.append(arr_rebinned)

    # compute covariance
    Cov = np.cov(arr_rebinned, rowvar=False)
    Cov_reg = Cov #+ eps * np.eye(new_bins)

    trace_ratios.append(np.trace(Cov_reg) / fid_trace)
    frob_ratios.append(np.linalg.norm(Cov_reg, "fro") / fid_frob)


trace_ratios = np.array(trace_ratios)
frob_ratios  = np.array(frob_ratios)

# plt.figure(figsize=(8,5))
# plt.scatter(np.arange(len(trace_ratios)), trace_ratios, s=15)
# plt.axhline(1, color='black', linestyle='--')
# plt.xlabel("Cosmology index")
# plt.ylabel("Trace ratio (rebinned)")
# plt.title("Covariance amplitude relative to fiducial")
# plt.show()

# plt.scatter(sigma8, trace_ratios)
# plt.axhline(1, color='black', linestyle='--')
# plt.xlabel("sigma8")
# plt.ylabel("Trace ratio")
# plt.show()

def rebin_2d(arr, new_bins):
    """
    arr: 2D array (N_real, old_bins)
    returns: (N_real, new_bins)
    """
    arr = np.asarray(arr, dtype=float)
    old_bins = arr.shape[1]
    factor = old_bins // new_bins
    trimmed = arr[:, :factor * new_bins]
    return trimmed.reshape(arr.shape[0], new_bins, factor).mean(axis=2)

fid_arr = np.asarray(cg_l1_kappa_realizations_fiducial[0], dtype=float)

if fid_arr.shape[1] != new_bins:
    fid_arr_reb = rebin_2d(fid_arr, new_bins)
else:
    fid_arr_reb = fid_arr


# ---------------------------
# Rebin theory means
# ---------------------------
theory = np.asarray(theory_l1_kappa_fix, dtype=float)   # (ncosmo, old_bins)

if theory.shape[1] != new_bins:
    theory_reb = rebin_2d(theory, new_bins)             # (ncosmo, new_bins)
else:
    theory_reb = theory

sim_reb = np.array(sim_reb, dtype=object)  # shape (ncosmo,), each is nreal×nbins

print("theory_reb shape:", theory_reb.shape)
print("length of sim_reb:", len(sim_reb))


# ----------------------------------------
# Generate samples for each cosmology
# ----------------------------------------
nsamples = 7

ncosmo, nbins = theory_reb.shape


samples_theory_fidcov = []   # list of nsamples × nbins for each cosmology
counter = 0
valid_cosmos = []
for i in range(ncosmo):
    if np.any(theory_l1_kappa_fix[i]< -1e-3):
        counter += 1
        valid_cosmos.append(i)
    mean_i = np.asarray(theory_reb[i], dtype=float)  # shape (nbins,)
    draws = np.random.multivariate_normal(mean=mean_i,
                                          cov=Cov_fid_reg,
                                          size=nsamples)
    samples_theory_fidcov.append(draws)

print("Skipped {} invalid cosmologies.".format(counter))

valid_cosmos = np.array(valid_cosmos)
samples_theory_fidcov = np.array(samples_theory_fidcov, dtype=object)  # shape (ncosmo,), each is 7×nbins

fig, axs = plt.subplots(1,2,figsize=(12,5))
axs[0].scatter(Om, sigma8, color='red', alpha=0.5)
axs[1].scatter(w_, h, color='green', alpha=0.5)
for i in valid_cosmos:
    axs[0].scatter(Om[i], sigma8[i], color='blue', alpha=0.5)
    axs[1].scatter(w_[i], h[i], color='blue', alpha=0.5)
axs[0].set_xlabel("Om")
axs[0].set_ylabel("sigma8")
axs[1].set_xlabel("w")
axs[1].set_ylabel("h")
```

```python
import numpy as np
import matplotlib.pyplot as plt


# =====================================================
# 1. Rebin function
# =====================================================
def rebin_2d(arr, new_bins):
    """
    Downsample the second axis of a 2D array by averaging contiguous blocks.

    Parameters
    ----------
    arr : array_like, shape (N_real, old_bins)
    new_bins : int

    Returns
    -------
    rebinned : ndarray, shape (N_real, new_bins)
    """
    arr = np.asarray(arr, dtype=float)
    old_bins = arr.shape[1]
    factor = old_bins // new_bins

    trimmed = arr[:, :factor * new_bins]
    return trimmed.reshape(arr.shape[0], new_bins, factor).mean(axis=2)


# =====================================================
# 2. Rebin fiducial and theory arrays
# =====================================================
fid_arr = np.asarray(cg_l1_kappa_realizations_fiducial[0], dtype=float)  # (N_real, old_bins)
fid_arr_reb = rebin_2d(fid_arr, new_bins) if fid_arr.shape[1] != new_bins else fid_arr

theory = np.asarray(theory_l1_kappa_fix, dtype=float)                    # (ncosmo, old_bins)
theory_reb = rebin_2d(theory, new_bins) if theory.shape[1] != new_bins else theory

sim_reb = np.asarray(sim_reb, dtype=object)

print("theory_reb shape:", theory_reb.shape)
print("length of sim_reb:", len(sim_reb))


# =====================================================
# 3. Draw multivariate samples for each cosmology
# =====================================================
nsamples = 7
ncosmo, nbins = theory_reb.shape

samples_theory_fidcov = []
invalid_cosmos = []

for i in range(ncosmo):

    # Skip cosmologies with negative PDF issues
    if np.any(theory_l1_kappa_fix[i] < -1e-3):
        invalid_cosmos.append(i)
        # continue

    mean_i = theory_reb[i]
    draws = np.random.multivariate_normal(mean=mean_i,
                                          cov=Cov_fid_reg,
                                          size=nsamples)
    samples_theory_fidcov.append(draws)

print(f"Skipped {len(invalid_cosmos)} invalid cosmologies.")
samples_theory_fidcov = np.array(samples_theory_fidcov, dtype=object)


# =====================================================
# 4. Visualise parameter space and invalid cosmologies
# =====================================================
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

axs[0].scatter(Om, sigma8, color='red', alpha=0.5, label="All")
axs[1].scatter(w_, h, color='green', alpha=0.5, label="All")

for idx in invalid_cosmos:
    axs[0].scatter(Om[idx], sigma8[idx], color='blue', alpha=0.5)
    axs[1].scatter(w_[idx], h[idx], color='blue', alpha=0.5)

axs[0].set_xlabel("Ωm")
axs[0].set_ylabel("σ8")
axs[1].set_xlabel("w₀")
axs[1].set_ylabel("h")
plt.show()


# =====================================================
# 5. Covariance amplitude comparison (trace, Frobenius)
# =====================================================
trace_ratios_samples = []
frob_ratios_samples = []

for draws in samples_theory_fidcov:

    arr = np.asarray(draws, dtype=float)  # (7, nbins)
    Cov_samp = np.cov(arr, rowvar=False)

    trace_ratios_samples.append(np.trace(Cov_samp) / fid_trace)
    frob_ratios_samples.append(np.linalg.norm(Cov_samp, "fro") / fid_frob)

trace_ratios_samples = np.array(trace_ratios_samples)
frob_ratios_samples = np.array(frob_ratios_samples)


# =====================================================
# 6. Plot amplitude vs cosmology index
# =====================================================
plt.figure(figsize=(10, 5))
plt.scatter(np.arange(ncosmo), trace_ratios, s=18, label="Simulation covariance")
plt.scatter(np.arange(len(samples_theory_fidcov)), trace_ratios_samples,
            s=18, label="Theory + FidCov samples")
plt.axhline(1, color="black", linestyle="--")
plt.xlabel("Cosmology index")
plt.ylabel("Trace ratio")
plt.legend()
plt.title("Simulation vs Theory+FidCov covariance amplitude")
plt.show()


# =====================================================
# 7. Scatter vs a parameter (example: sigma8)
# =====================================================
plt.figure(figsize=(7, 5))
plt.scatter(sigma8, trace_ratios, s=18, label="Sim")
plt.scatter(sigma8, trace_ratios_samples, s=18, label="Theory+FidCov")
plt.axhline(1, color="black", linestyle="--")
plt.xlabel("σ8")
plt.ylabel("Trace ratio")
plt.legend()
plt.show()
```
